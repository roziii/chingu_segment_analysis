```python
import pandas as pd 
import numpy as np 
import warnings
warnings.filterwarnings('ignore')
```

```python
voyage_lists = pd.read_csv('./data/DonationAnalysis-Chingu_VoyageSchedules_20250305.csv')
applications = pd.read_csv('./data/RedactedDonationAnalysis-Chingu_Applications_20250304.csv')
completion = pd.read_csv('./data/RedactedDonationAnalysis-Chingu_VoyageCompletions_20250305 - DonationAnalysis-Chingu_VoyageCompletions_20250305.csv')
transactions = pd.read_csv('./data/RedactedDonationAnalysis-Chingu_RevenueTxns_20250305.csv')
```

```python
# Extract list of unique voyages from schedules
voyage_list = voyage_lists["Name"].dropna().unique()

# Merge data on Unique ID
merged_df = applications.merge(transactions, on="Unique ID", how="left")
merged_df = merged_df.merge(completions[["Unique ID", "What is your Voyage?"]], on="Unique ID", how="left")

# Create voyage participation columns
for voyage in voyage_list:
    merged_df[voyage] = (merged_df["What is your Voyage?"] == voyage).astype(int)

# Drop redundant voyage column
merged_df.drop(columns=["What is your Voyage?"], inplace=True)

# Display the final merged dataset
import ace_tools as tools
tools.display_dataframe_to_user(name="Merged Voyage Data", dataframe=merged_df)
```
