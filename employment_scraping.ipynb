```python
from urllib.request import urlopen, Request
from bs4 import BeautifulSoup
url = 'https://www.bls.gov/charts/employment-situation/civilian-unemployment-rate.html'
req = Request(url)

html = urlopen(req).read()
bs = BeautifulSoup(html, 'html.parser')

# Find all rows
rows = bs.find_all('tr')

for row in rows:
    # Extract date from <p class="sub0">
    date_tag = row.find('p', class_='sub0')
    # Extract value from <span class="datavalue">
    value_tag = row.find('span', class_='datavalue')

    if date_tag and value_tag:
        date = date_tag.get_text(strip=True)
        value = value_tag.get_text(strip=True)
        print(f"{date}: {value}")
```
